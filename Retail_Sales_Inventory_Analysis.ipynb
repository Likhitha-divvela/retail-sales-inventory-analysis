```python
sales_clean = sales.copy()
inventory_clean = inventory.copy()
```

```python
sales_clean = sales_clean.drop(columns=["Reason of Return"])
sales_clean.shape
```

```python
sales_text_columns = sales_clean.select_dtypes(include="string").columns

for col in sales_text_columns:
    sales_clean[col] = sales_clean[col].str.strip()

inventory_text_columns = inventory_clean.select_dtypes(include="string").columns

for col in inventory_text_columns:
    inventory_clean[col] = inventory_clean[col].str.strip()
```

```python
#Gross Sales
sales_clean["Gross Sales"] = sales_clean["Sales Amount"].clip(lower=0)

#Retuen Amount
sales_clean["Return Amount"] = sales_clean["Sales Amount"].clip(upper=0).abs()

#Net Sales
sales_clean["Net Sales"] = sales_clean["Sales Amount"]

#Profit
sales_clean["Profit"] = (
    sales_clean["Sales Amount"] - sales_clean["Cogs"]
)

#Profit Margin 
sales_clean["Profit Margin"] = np.where(
    sales_clean["Net Sales"] != 0,
    sales_clean["Profit"] / sales_clean["Net Sales"],
    0
)

#Month
sales_clean["Month"] = sales_clean["Transaction Date"].dt.to_period("M").astype(str)

#Year 
sales_clean["Year"] = sales_clean["Transaction Date"].dt.year

#Month Name
sales_clean["Month Name"] = sales_clean["Transaction Date"].dt.strftime("%B")

#Inventory Value
inventory_clean["Inventory Value"] = inventory_clean["Stocks Selling Amount"]

#Inventory Cost
inventory_clean["Inventory Cost"] = inventory_clean["Cost of Stocks"]

#Stock availability flag
inventory_clean["Stock Available"] = np.where(
    inventory_clean["Qty on hand"] > 0,
    "In Stock",
    "Out of Stock"
)

#Negative inventory flag
inventory_clean["Negative Stock Flag"] = np.where(
    inventory_clean["Qty on hand"] < 0,
    "Negative Stock",
    "Normal"
)

```

```python
total_gross_sales = sales_clean["Gross Sales"].sum()
total_returns = sales_clean["Return Amount"].sum()
net_sales = sales_clean["Net Sales"].sum()
total_cogs = sales_clean["Cogs"].sum()
total_profit = sales_clean["Profit"].sum()
total_units = sales_clean["Qty Sold"].sum()
total_transactions = sales_clean["Number of Transactions"].sum()

profit_margin = total_profit / net_sales
return_rate = total_returns / total_gross_sales

print("Total Gross Sales:", round(total_gross_sales, 2))
print("Total Returns:", round(total_returns, 2))
print("Net Sales:", round(net_sales, 2))
print("Total COGS:", round(total_cogs, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Units Sold:", round(total_units, 2))
print("Total Transactions:", total_transactions)
print("Profit Margin:", round(profit_margin * 100, 2), "%")
print("Return Rate:", round(return_rate * 100, 2), "%")
```

```python
#Create stock-risk indicator
stockout_risk = product_sales_inventory[
    (product_sales_inventory["Units_Sold"] > 0) &
    (product_sales_inventory["Current_Inventory_Units"] <= 0)
].copy()

stockout_risk.sort_values(
    "Units_Sold",
    ascending=False
).head(20)
```

```python
product_sales_inventory["Inventory_to_Sales_Ratio"] = np.where(
    product_sales_inventory["Net_Sales"] > 0,
    product_sales_inventory["Current_Inventory_Value"] /
    product_sales_inventory["Net_Sales"],
    np.nan
)

product_sales_inventory.sort_values(
    "Inventory_to_Sales_Ratio",
    ascending=False
).head(20)
```

```python
#Stock-out Risk Analysis
stockout_risk = product_sales_inventory[
    (product_sales_inventory["Units_Sold"] > 0) &
    (product_sales_inventory["Current_Inventory_Units"] <= 0)
].copy()

stockout_risk = stockout_risk.sort_values(
    "Units_Sold",
    ascending=False
)

stockout_risk.head(20)
print("Potential stock-out products:", len(stockout_risk))

print(
    "Net Sales at stock-out risk:",
    round(stockout_risk["Net_Sales"].sum(), 2)
)

print(
    "Units Sold from stock-out risk products:",
    stockout_risk["Units_Sold"].sum()
)
```

```python
#Inventory Coverage
sales_months = 11

product_sales_inventory["Avg_Monthly_Units_Sold"] = (
    product_sales_inventory["Units_Sold"] / sales_months
)
product_sales_inventory["Inventory_Coverage_Months"] = np.where(
    product_sales_inventory["Avg_Monthly_Units_Sold"] > 0,
    product_sales_inventory["Current_Inventory_Units"] /
    product_sales_inventory["Avg_Monthly_Units_Sold"],
    np.nan
)
high_inventory_coverage = (
    product_sales_inventory[
        product_sales_inventory["Current_Inventory_Units"] > 0
    ]
    .sort_values(
        "Inventory_Coverage_Months",
        ascending=False
    )
)

high_inventory_coverage.head(20)
```

```python
#Risk Classification
def classify_inventory_risk(row):
    if row["Current_Inventory_Units"] <= 0 and row["Units_Sold"] > 0:
        return "Stock-out Risk"
    elif row["Inventory_Coverage_Months"] >= 6:
        return "Potential Overstock"
    elif row["Inventory_Coverage_Months"] <= 1:
        return "Low Stock"
    else:
        return "Healthy"

product_sales_inventory["Inventory_Risk"] = (
    product_sales_inventory.apply(
        classify_inventory_risk,
        axis=1
    )
)
```
